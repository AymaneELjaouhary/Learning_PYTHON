```python
# Our execise is to make a bar chart for job title's median salary

#import labraries
import pandas as pd
import matplotlib.pyplot as plt

# Data & Fix the date type
df = pd.read_csv (r"C:\Users\PC\Documents\PYTHON\data_jobs.csv")
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df.info()
```

```python
# Make the median salary
median_salary = df.groupby ('job_title_short')['salary_year_avg'].median().sort_values()
median_salary
```

```python
# make the BarH chart
plt.barh (median_salary.index , median_salary)
plt.title ('Median Salary by Job Title')
plt.xlabel ('Salary ($USD)')
plt.show()
```
