```python
sales_categories = df.groupby( ['product_category'])['price'].count().sort_values()
sales_categories
```

```python
Total_sales = sales_categories.sum()
Percentage_sales = []
for x in sales_categories :
    result = (x/Total_sales)*100
    Percentage_sales.append(result)

Percentage_sales
```
