```python
# Import Labraries
import pandas as pd
import matplotlib.pyplot as plt
import ast

# Data & Cleaning
df = pd.read_csv(r'C:\Users\PC\Documents\Learning_PYTHON\data_jobs.csv')
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply ( lambda x: ast.literal_eval(x) if pd.notna(x) else x)

# in this lessons , You will learn how to do the 'Pie Chart'
```

```python
df.info()
```

```python
# Introduction 
df['job_work_from_home'].value_counts().plot(kind="pie" , startangle = 90 , autopct = '%1.1f%%')
plt.title ("Work from Home Status")
plt.show()

# Congratulations 
```

```python
# Now Let's use it in multiple chart in one time 
# Use the 3 booliean columns in the data for that 

df[['job_health_insurance' , 'job_no_degree_mention' , 'job_work_from_home']]
```

```python
fig,ax = plt.subplots(1,3)

dict_names = {
    'job_work_from_home' : 'Work from home' ,
    'job_health_insurance' : 'Work Insurance' , 
    'job_no_degree_mention' : 'Job Degree' ,
}

for i,(column,title) in enumerate(dict_names.items()) : 
    ax[i].pie(df[column].value_counts() , startangle=90 , autopct='%1.1f%%' , labels = ['False' , 'True'])
    ax[i].set_title(title)

```
