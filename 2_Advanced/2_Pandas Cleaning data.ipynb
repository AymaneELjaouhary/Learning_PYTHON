```python
# import Librairies
import pandas as pd 
import matplotlib.pyplot as plt

# Data & Clean
df = pd.read_csv(r'C:\Users\PC\Documents\Learning_PYTHON\data_jobs.csv')
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df.info()
```

```python
df.loc [:10 , 'salary_rate':'salary_hour_avg']
```

```python
# Our job in this stage is to replace all the NAN salary values by median salary values , and Drop duplicate 

# Calculate the median
median_salary_hour = df['salary_hour_avg'].median()
median_salary_year = df['salary_year_avg'].median()
```

```python
# Fill the NaN values by the median using " fillna() "
df['salary_hour_avg'] = df['salary_hour_avg'].fillna(median_salary_hour)
df['salary_year_avg'] = df['salary_year_avg'].fillna(median_salary_year)
```

```python
# Result
df_filled = df
df_filled.loc[:10 , 'salary_year_avg':'salary_hour_avg']

```

```python
# Drop Duplicates , Using " .drop_duplicate "
df_unique = df_filled
df_unique = df_unique.drop_duplicates()

print ('Length of original df is :      ' , len(df_filled) )
print ('Lenght of drop duplicate is :   ' , len(df_unique))
print ( 'Rows dropped :                  ' , len(df_filled)-len(df_unique))
```

```python
# Now lets drop duplicates but with 2 column not the entire data 
df_unique_2 = df_unique.drop_duplicates (subset =['job_title' , 'company_name'])

print ('Length of original df is :      ' , len(df_unique) )
print ('Lenght of drop duplicate is :   ' , len(df_unique_2))
print ( 'Rows dropped :                  ' , len(df_unique)-len(df_unique_2))
```
