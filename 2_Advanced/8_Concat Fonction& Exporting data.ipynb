```python
# import Librairies
import pandas as pd 
import matplotlib.pyplot as plt

# Data & Clean
df = pd.read_csv(r'C:\Users\PC\Documents\Learning_PYTHON\data_jobs.csv')
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)

# Your mession is to add a data to other data ( add rows and not the columns )
# We will split our data by months , so we can concat the data month by month
```

```python
df['job_posted_month'] = df['job_posted_date'].dt.strftime('%b')  # Using %b to show only the first 3 letter of the month like jan
```

```python
# Split the data into 12 months . And the easieat way to do that is to make a list 

months = df['job_posted_month'].unique()

dict_months = {month : df[df['job_posted_month'] == month] for month in months } 

dict_months['Jan'].head()
```

```python
# Concat Quarter 1

df_q1 = pd.concat([  dict_months['Jan'] , dict_months['Feb'] , dict_months['Mar']  ] , ignore_index =True)
df_q1.value_counts()
```
